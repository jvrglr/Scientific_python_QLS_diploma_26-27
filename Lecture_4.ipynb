## Hands on V: Branches, checkout and merge

So far we discussed a sequential workflow where we use the structure modify->add-> commit, and synchronize with the remote repository through push and pull. One might want to alter such workflow, for example going to a previous state of the project to solve bugs or to add new features to the code.

Branches allows us to create new instances of our code and work on them independently to the state of files in the main branch. Branches are created through **git branch branch_name** where we indicate the name of the new branch. Create a new branch called "test-branch". Then list all existing branches with **git branch**

```python
git branch test-branch # Create a new branch called "test-branch"
git branch # List all branches in the repository
```

We switch between branches with **git checkout**

```python
git checkout test-branch # Switch to the "test-branch" branch
```
