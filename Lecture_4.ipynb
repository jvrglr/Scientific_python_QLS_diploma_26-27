## Matplotlib

When science meets arts, let's plot!

```python
import numpy as np
import matplotlib.pyplot as plt
```

Basic commands I: scatter

```python
xs = np.linspace(-5, 5, 100) # x values (independent variable) 
ys = np.sin(xs) # y values (dependent variable)
plt.scatter(xs, ys)  #Plot data as a scatter plot
```

Basic commands II: plot

```python
plt.plot(xs,ys) #Plot data joined by lines
```

Plotting multiple things together

```python
#Plot the same data as a scatter plot and a line plot
plt.plot(xs,ys)
plt.scatter(xs,ys)
```

MUST DO: Naming labels 

```python
plt.plot(xs,ys)
plt.ylabel("sin x") #y-axis label
plt.xlabel("x") #x-axis label
```

Formatting I: Choosing colors and size

```python
plt.plot(xs,ys, color = "grey", lw = 2) # lw = line width
plt.scatter(xs,ys, color = "red", s = 30 ) #s = size of the points
plt.ylabel("sin x")
plt.xlabel("x")
```

Formatting III: Colors can be chosen from a colormap (maps numbers from 0 to one to a given color in a palette): https://matplotlib.org/stable/users/explain/colors/colormaps.html

```python
n_curves = 6 # number of curves to plot and colors to generate

cmap = plt.colormaps["GnBu"]
numbers = np.linspace(0, 1, n_curves)  # Generate evenly spaced numbers between 0 and 1
colors = cmap(numbers)  # Convert numbers to colors using the colormap

x = np.linspace(0, 10, 200)
for i, c in enumerate(colors):
    y = np.sin(x + i * 0.5)
    plt.plot(x, y, color=c)
```

Basic commands III: Setting figure size and multiple plots

```python
#Create data to plot
y1 = np.sin(xs)
y2 = np.sin(xs + 0.2)
y3 = np.sin(xs + 0.4)
y4 = np.sin(xs + 0.6)
y5 = np.sin(xs + 0.8)
y6 = np.sin(xs + 1)

# now we decide the actual figure size in inches
fig = plt.figure(figsize=(12,6))

#plt.subplot(ijk) where i is the number of rows, j is the number of columns, and k is the index of the plot (k =1,2,3,...,i*j)
#--------------First plot: Row 1, Column 1
plt.subplot(231)
plt.ylabel("sin x")
plt.xlabel("x")
plt.plot(xs,y1,color='r')
#--------------Second plot:
plt.subplot(232)
plt.plot(xs,y2,color='g')
#--------------Third plot:
plt.subplot(233)
plt.plot(xs,y3,color='b')
#--------------Fourth plot:
plt.subplot(234)
plt.plot(xs,y4,color='y')
#--------------Fifth plot:
plt.subplot(235)
plt.plot(xs,y5,color='c')
#--------------Sixth plot:
plt.subplot(236)
plt.plot(xs,y6,color='m')
```

Formatting III: Choosing transparency and z-order

```python
fig = plt.figure(figsize=(16,4))

plt.subplot(131)
plt.title("Line behind points")
plt.plot(xs,ys, color = "grey", lw = 2, zorder = 0)
plt.scatter(xs,ys, color = "red", s = 30 , zorder = 1 )

plt.subplot(132)
plt.title("Line on top of points")
plt.plot(xs,ys, color = "grey", lw = 2, zorder = 1)
plt.scatter(xs,ys, color = "red", s = 30 , zorder = 0 )

plt.subplot(133)
plt.title("Line and points with transparency")
plt.plot(xs,ys, color = "grey", lw = 2, alpha = 0.5) #Transparency of the line controlled by alpha parameter, alpha \in [0,1]
plt.scatter(xs,ys, color = "red", s = 30 , alpha = 0.5 )
```

Formatting IV: Markers and linestyles

```python
fig = plt.figure(figsize=(16,4))
plt.plot(xs,y1,linestyle='-',marker=('o'))
plt.plot(xs,y2,linestyle='--',color='#00FA9A',marker=('^'))
plt.plot(xs,y3,linestyle='-.',color=(139/255, 0, 139/255),marker=('s'))
plt.plot(xs,y4,linestyle=':',color='lightcoral',marker=('*'))
plt.plot(xs,y5,linestyle=' ',color='steelblue',marker=('$H$'))
```

Formatting V: Fontsize and math mode

```python
ws = np.array([np.pi/2, np.pi])
ws_string = [r'$\omega = \frac{\pi}{2}$', r'$\omega = \pi$']
for ii,w in enumerate(ws):
    yw = np.sin(xs+w)
    plt.plot(xs, yw, label=ws_string[ii])
plt.legend(fontsize=12, loc='upper right')
```

Formatting VI: LaTeX font

```python
#-----------You just run this once to set the font for all the plots in the script/notebook----------------
plt.rcParams['mathtext.fontset'] = 'cm'
plt.rcParams['font.family'] = 'serif'
plt.rcParams['text.usetex'] = True
plt.rcParams['font.family'] = 'serif'
#----------------------------------------------------------------------------------------------------------
for ii,w in enumerate(ws):
    yw = np.sin(xs+w)
    plt.plot(xs, yw, label=ws_string[ii])
plt.legend(fontsize=12, loc='upper right')
```

Basic commands IV: Making histograms

```python
xs = np.random.normal(0, 5, 100) # Random data
n_bins = 10 # Number of bin edges
bins = np.linspace(min(xs), max(xs), n_bins) # Bin edges

fig = plt.figure(figsize=(16,4))
plt.subplot(121)
plt.title("Non-normalized histogram (frequencies)")
plt.hist(xs, bins=bins, color='lightblue', edgecolor='black')
plt.subplot(122)
plt.title("Normalized histogram (probability density)")
plt.hist(xs, bins=bins, color='lightblue', edgecolor='black', density=True)
```

VERY IMPORTANT! ALL FIGURES OF THE COURSE (and of any scientific document) HAVE TO FULFILL THESE RULES:
1. All data is visible.
2. All text is legible (same fontsize as in the main text if any).
3. Information of what is plotted is either in legends or in the figure caption.
4. Use different linestyles, markers and colors to differentiate different curves and data. NOT ONLY DIFFERENT COLORS.

## Pandas

Dealing with data.

```python
import pandas as pd
```

Generate a DataFrame from a dictionary: Keys are column names and values are the content of the columns.

```python
#Create a dictionary with data
a_dictionary = {
  'cat_names': ["Milo", "Gordo", "Luna"],
  'cats': ["grey", "black", "white"],
  'age': [3, 7, 2]
}

df_cats = pd.DataFrame(a_dictionary) #Create a DataFrame from the dictionary
```

Basic commands I: See content of dataframe

```python
df_cats.head(3) # df.head(n) returns the first n rows of the DataFrame, default is 5
```

Navigating a dataframe I: Display column content

```python
df_cats["age"] #We can access columns through the column name.
```

You can operate with columns. Operations are not stored by default.

```python
df_cats["age"] + 1 # This is not storing the result in the DataFrame, it is just returning a new Series with the result of the operation.
print(df_cats.head())
print("\n")
df_cats["age"] = df_cats["age"] + 2 # This is storing the result in the DataFrame in the column "age"
print(df_cats.head())
```

Navigating a dataframe II: Display rows with default label

```python
df_cats.iloc[0] #To access a row with an integer index "row_index", we use .iloc[row_index]
```

By default, rows are labeled with integers, but we could also name rows just as we do with columns

```python
df_cats.index = ['cat 1', 'cat 2', 'cat 3']
df_cats.head()
```

Navigating a dataframe III: Display rows with changed label iloc vs loc

```python
print("Using loc:")
print(df_cats.loc['cat 1']) # Accessing row by label
print("\n")
print("Using iloc:")
print(df_cats.iloc[0]) # Accessing row by index
```

When we open a .csv file with pandas by default rows are labeled with an integer, we can instead use any of the columns to define the index names

```python
df = pd.read_csv("Data/people-10.csv", index_col = "First Name")
df.head()
```

Manipulating data I: From DataFrame to python

```python
ages = df["Age"].to_numpy() # Creates a numpy array from the column "Age" of the DataFrame
print(ages)
```

Manipulating data II: Adding columns

```python
twice_ages = 2*ages
df["Twice Age"] = twice_ages
df.head()
```

Manipulating data III: Removing columns

```python
df = df.drop(columns=["Email", "Twice Age"])
df.head() 
```

Manipulating data IV: Filtering data by column value

```python
df_filtered = df[df["Sex"]=="Male"] # Returns a DataFrame with only the rows where the column "Sex" is equal to "Male"
df_filtered.head()
```

Manipulating data V: Conditioned operations (make distinct operation on column elements depending on value of a different column). As an example, we add a column in which we add +10 to the age if the subject is Male.

```python
import numpy as np
df["new column"] = df["Age"] #Duplicates the column "Age" and creates a new column called "new column"
df["new column"] = np.where(df["Sex"]=="Male", df["new column"] +10, df["new column"] ) #Creates a new column called "new column" with the values of the column "Age" if the column "Sex" is equal to "Male", and keep the same values otherwise
df.head()
```

Manipulating data VI: Creating a smaller dataset with just a few columns

```python
df_small = df[["Sex", "Age"]].copy() # Creates a new DataFrame with only the columns "Sex" and "Age" from the original DataFrame
df_small.head()
```

Manipulating data VI: Grouping data by column value. When grouping we usually select a grouping operation, typical ones are:
1. sum()--> Add values
2. mean() --> Average values
3. std() --> Standard deviation
4. count() --> number of occurrences

```python
df_grouped = df_small.groupby("Sex").count().reset_index() # Groups the DataFrame by the column "Sex" and counts the number of rows in each group, then resets the index to have a clean DataFrame
df_grouped.head()
```

DataFrames can have missing data. Pandas uses the value NaN (Not a Number) to represent missing data. 

```python
df = pd.read_csv("Data/people-10_missing_values.csv", index_col = "Index", na_values=["N/A", "Unknown"])
df.head()
```

Clean data I: remove all rows with, at least, one empty entry

```python
cleaned_df = df.dropna() #returns a new dataframe
cleaned_df.info()
cleaned_df.head()
```

Clean data II: Replace missing data

```python
df = pd.read_csv("Data/people-10_missing_values.csv") #Load data again to show how to replace missing values
df = df.fillna("Missing") #Replaces ALL missing values
df.head()
```
