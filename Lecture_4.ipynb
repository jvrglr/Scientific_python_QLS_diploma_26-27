## Matplotlib

When science meeets arts, let's plot!

```python
import numpy as np
import matplotlib.pyplot as plt
```

Basic commands I: scatter

```python
xs = np.linspace(-5, 5, 100) # x values (independent variable) 
ys = np.sin(xs) # y values (dependent variable)
plt.scatter(xs, ys) 
```

Basic commands II: plot

```python
plt.plot(xs,ys)
```

Plotting multiple things together

```python
plt.plot(xs,ys)
plt.scatter(xs,ys)
```

MUST DO: Naming labels 

```python
plt.plot(xs,ys)
plt.ylabel("sin x")
plt.xlabel("x")
```

Formatting I: Choosing colors, and size

```python
plt.plot(xs,ys, color = "grey", lw = 2)
plt.scatter(xs,ys, color = "red", s = 30 )
plt.ylabel("sin x")
plt.xlabel("x")
```

Basic commands III: Setting figure size and multiple plots

```python
#Create data to plot
y1 = np.sin(xs)
y2 = np.sin(xs + 0.2)
y3 = np.sin(xs + 0.4)
y4 = np.sin(xs + 0.6)
y5 = np.sin(xs + 0.8)
y6 = np.sin(xs + 1)

# now we decide the actual figure size in inches
fig = plt.figure(figsize=(12,6))

#plt.subplot(ijk) where i is the number of rows, j is the number of columns, and k is the index of the plot (k =1,2,3,...,i*j)
#--------------First plot: Row 1, Column 1
plt.subplot(231)
plt.ylabel("sin x")
plt.xlabel("x")
plt.plot(xs,y1,color='r')
#--------------Second plot:
plt.subplot(232)
plt.plot(xs,y2,color='g')
#--------------Third plot:
plt.subplot(233)
plt.plot(xs,y3,color='b')
#--------------Fourth plot:
plt.subplot(234)
plt.plot(xs,y4,color='y')
#--------------Fifth plot:
plt.subplot(235)
plt.plot(xs,y5,color='c')
#--------------Sixth plot:
plt.subplot(236)
plt.plot(xs,y6,color='m')
```

Formatting II: Choosing transparency, and z-order

```python
fig = plt.figure(figsize=(16,4))

plt.subplot(131)
plt.title("Line behind points")
plt.plot(xs,ys, color = "grey", lw = 2, zorder = 0)
plt.scatter(xs,ys, color = "red", s = 30 , zorder = 1 )

plt.subplot(132)
plt.title("Line on top of points")
plt.plot(xs,ys, color = "grey", lw = 2, zorder = 1)
plt.scatter(xs,ys, color = "red", s = 30 , zorder = 0 )

plt.subplot(133)
plt.title("Line and points with transparency")
plt.plot(xs,ys, color = "grey", lw = 2, alpha = 0.5) #Transparency of the line controlled by alpha parameter, alpha \in [0,1]
plt.scatter(xs,ys, color = "red", s = 30 , alpha = 0.5 )
```

Formatting IV: Markers and linestyles

```python
fig = plt.figure(figsize=(16,4))
plt.plot(xs,y1,linestyle='-',marker=('o'))
plt.plot(xs,y2,linestyle='--',color='#00FA9A',marker=('^'))
plt.plot(xs,y3,linestyle='-.',color=(139/255, 0, 139/255),marker=('s'))
plt.plot(xs,y4,linestyle=':',color='lightcoral',marker=('*'))
plt.plot(xs,y5,linestyle=' ',color='steelblue',marker=('$H$'))
```

Formatting V: Fontsize and math mode

```python
ws = np.array([np.pi/2, np.pi])
ws_string = [r'$\omega = \frac{\pi}{2}$', r'$\omega = \pi$']
ws = np.array([np.pi/2, np.pi])
ws_string = [r'$\omega = \frac{\pi}{2}$', r'$\omega = \pi$']
for ii,w in enumerate(ws):
    yw = np.sin(xs+w)
    plt.plot(xs, yw, label=ws_string[ii])
plt.legend(fontsize=12, loc='upper right')
```

Formatting VI: Latex font

```python
#-----------You just run this once to set the font for all the plots in the script/notebook----------------
plt.rcParams['mathtext.fontset'] = 'cm'
plt.rcParams['font.family'] = 'serif'
plt.rcParams['text.usetex'] = True
plt.rcParams['font.family'] = 'serif'
#----------------------------------------------------------------------------------------------------------
for ii,w in enumerate(ws):
    yw = np.sin(xs+w)
    plt.plot(xs, yw, label=ws_string[ii])
plt.legend(fontsize=12, loc='upper right')
```

## Pandas

Dealing with data.

```python
import pandas as pd
```

Generate a dataset from a dictionary: Keys are column names and values are the content of the columns.

```python
a_dictionary = {
  'cat_names': ["Milo", "Gordo", "Luna"],
  'cats': ["grey", "black", "white"],
  'age': [3, 7, 2]
}

df_cats = pd.DataFrame(a_dictionary)
```

Basic commands I: See content of dataframe

```python
df_cats.head(3) # df.head(n) returns the first n rows of the DataFrame, default is 5
```

Navigating a dataframe I: Display columns

```python
df_cats["age"] #We can access columns through the column name.
```

Navigating a dataframe II: Display rows with deafault label

```python
df_cats.iloc[0] #To access a row, we use .loc[row_index]
```

By deafault, rows are labeled with integers, but we could also name rows just as we do with columns

```python
df_cats.index = ['cat 1', 'cat 2', 'cat 3']
df_cats.head()
```

Navigating a dataframe III: Display rows with changed label iloc vs loc

```python
print("Using loc:")
print(df_cats.loc['cat 1']) # Accesing row by label
print("\n")
print("Using iloc:")
print(df_cats.iloc[0]) # Accesing row by index
```

When we open a .csv file with pandas by deafoult rows are labeled with an integer, we can rather use any of the columns to define the index names

```python
df = pd.read_csv("Data/people-10.csv", index_col = "First Name")
df.head()
```

Manipulating data I: From DataFrame to python

```python
ages = df["Age"].to_numpy()
print(ages)
```

Manipulating data II: Adding columns

```python
twice_ages = 2*ages
df["Twice Age"] = twice_ages
df.head()
```

Manipulating data III: Removing columns

```python
df = df.drop(columns=["Email", "Twice Age"])
df.head() 
```

DataFrames can have missing data. Pandas uses the value NaN (Not a Number) to represent missing data. 

```python
df = pd.read_csv("Data/people-10_missing_values.csv", index_col = "Index", na_values=["N/A", "Unknown"])
df.head()
```

```python
df.info()
```

Clean data I: remove all rows with, at least, one empty entry

```python
cleaned_df = df.dropna() #returns a new dataframe
cleaned_df.info()
cleaned_df.head()
```

Clean data II: Replace missing data

```python
df = pd.read_csv("Data/people-10_missing_values.csv") #Load data again to show how to replace missing values
df = df.fillna("Missing") #Replaces ALL missing values
df.head()
```

```python
df = pd.read_csv("Data/people-10_missing_values.csv")  #Load data again to show how to replace missing values
df["Age"] = df["Age"].fillna(30) #Replaces ALL missing values
df["Age"]
```
