## Virtual environments
Different projects might need different packages and different **versions** of them. Environments help us making our code robust to undesired package updates, also enhances portability,

Creating a virtual environment with Conda:

```python
conda create -n myenv
```

Activate a virtual environment:

```python
conda activate myenv
```

Install a package in an activated environment:

```python
conda install packagename
```

We can specify what version of our package we want to install (this can be important for importing the code of others)

```python
conda install numpy=2.2.2
```

Install Jupyter

```python
conda install jupyter
```

Alternatives to Conda: use the official distribution (pip)

## Git setup

Setting minimal configuration: git needs to know who is executing changes in the code, therefore we have to identify ourselves at least once providing our name and email.

Basic tutorial: https://www.w3schools.com/git/default.asp?remote=github

```python
git config --global user.name "Your Name" # set name
git config --global user.email "your.email@example.com" # set email
git config --list # see the list of settings
```

## Our first repository -- basic git commands
A Git repository is a folder that Git tracks for changes.

We will use git in the folder we created for our "Hello, World!" program.

We have to create a repository through the command **git init**.  Basically it creates a hidden repository (.git) where all data of changes is stored. This should be the first git command.

```python
git init # Run this in the root directory of your project
ls -a # Check if .git folder is created
```

As our workflow progress, the state of the different files of the folder will change for git. Files can be
1. Added and commited (git is fully aware of them and all changes were registered)
2. Added and not commited (git is aware of the file, but there are changes not stored in git)
3. Untracked (git is not aware of the file)

**git status** allows us to see the state of the files in our repository. Now our only file should be untracked.

```python
git status # Check the status of your repository
```

Git starts tracking files when we first **add** them

```python
git add first_program.py #Start tracking the file
git status # Check how the status of your repository changed
```

Some times we want to add several files at the time, we can use

```python
git add -A # Add all files in the repository
git add . # Add all files in the current folder and its subfolders (not parent folders)
```

We can undo the add command if we staged a file by mistake. This doesn't remove the file, only tells git to stop tracking the last changes.

```python
git restore --staged filename # Unstage a file
```

Staged/added files are not really stored by git. Adding only serves to tell git what we actually want to save. **commit** is the command used to save snapshots of our project. The commit program **always** comes with a message specifying what were the changes from the last commit.

```python
git commit -m "Initial commit" # Commit the changes with a message
git status # Check how the status of the repository changed.
```

The history of commits is tracked, and we can always go back to a previous commit. To see the record of commits we use git log

![gitlog](Images/Screenshot_git_Log.png)

```python
git log # View the commit history
```
