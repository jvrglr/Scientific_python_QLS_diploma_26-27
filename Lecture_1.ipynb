## Anaconda / Conda
Leading plataform for python. It would help us managing Python packages and creating virtual environments. "*Conda handles installing software, resolving dependencies, and keeping your projects isolated from each other so that updating one project never breaks another.*"

Documentation: https://www.anaconda.com/docs/main 


Download Anaconda from https://www.continuum.io/downloads


## Virtual environments
Different projects might need different packages and different **versions** of them. Environments help us making our code robust to undesired package updates, also enhances portability,

Creating a virtual environment with Conda:

```python
conda create -n myenv
```

Activate a virtual environment:

```python
conda activate myenv
```

Install a package in an activated environment:

```python
conda install packagename
```

We can specify what version of our package we want to install (this can be important for importing the code of others)

```python
conda install numpy=2.2.2
```

Install Jupyter

```python
conda install jupyter
```

Alternatives to Conda: use the official distribution (pip)

# Preparing the elements of our first program: Print and comment

print: function prints the specified message to the screen. Here we will use a python string to write the message.

```python
print("Hello, World!")
```

Comments are preceded by a "#" character, these messages are not read by python, it serves us to explain the code.

## My first program; Python scripts and Jupyter notebook cells.

 The following instructions use VS code, but you can use any text editor of your choice.

 Let us write our first Python code using print. We will run the code both as a script and using jupyter cells.

1. Create a new folder for the lesson.
2. Open the folder with VS Code.
2. Inside the folder, create a new file named `first_program.py`.
3. Open the `first_program.py` file in a text editor like VS Code.
4. Add a print and a comment with two distinct messages each.
5. Run the script by opening a terminal in VS Code and typing `python first_program.py` (or `python3 first_program.py` depending on your Python installation). The message used as argument of the print function should appear, but the comment shouldn't be shown.


![Open terminal](Images/Screenshot_VS_code_open_terminal.png)


Now we will repeat the same thing but using cells. 

In VS code it is particularly easy to jump between the scripts and cells. We will use a particular comment, which is ignored by python but will tell VS code that we want to enable the interactive window for Jupyter cells.

1. Open the script `first_program.py` that we created before in VS code.
2. Add the comment "# %%" in the first line.
3. Adding the comment above should have enabled cell options (Jupyter extension must be installed).
4. Run the code as a cell using Shift+Enter. The cell should produce the same outcome as when we run `python first_program.py` in the terminal.
5. Create a new cell below the firt one and add a second print.
6. Write a third print instance in the interactive window. Note that in the interactive window, code is run pressing Enter while change on line occurs with Shift+Enter (the oposite that in cells).
7. Run again `python first_program.py` in the terminal. Do you understand how the output of the terminal, cells and the interactive window works?


Documentation Python interactive window: https://code.visualstudio.com/docs/python/jupyter-support-py

**Warning**: Cells and the interactive window are very useful, they allow testing things and looking for bugs quite nicely as we divide the code in sub-cells that can be executed independently. BUT cells are also dangerous, remember that you can run independent cells at arbitrary orders but python scripts, when run in the terminal always go sequentially, from the begining of the file to the bottom.
