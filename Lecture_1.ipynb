# Setup: Anaconda and (virtual) environments

## Anaconda / Conda
Leading plataform for python. It would help us managing Python packages and creating virtual environments. "*Conda handles installing software, resolving dependencies, and keeping your projects isolated from each other so that updating one project never breaks another.*"

Documentation: https://www.anaconda.com/docs/main 


Download Anaconda from https://www.continuum.io/downloads


## Virtual environments
Different projects might need different packages and different **versions** of them. Environments help us making our code robust to undesired package updates, also enhances portability,

Creating a virtual environment:

```python
conda create -n myenv python
```
