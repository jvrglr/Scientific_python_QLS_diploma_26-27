## Our first repository -- basic git commands
A Git repository is a folder that Git tracks for changes.

We will use git in the folder we created for our "Hello, World!" program.

We have to create a repository through the command **git init**.  Basically it creates a hidden repository (.git) where all data of changes is stored. This should be the first git command.

```python
git init # Run this in the root directory of your project
ls -a # Check if .git folder is created
```

As our workflow progress, the state of the different files of the folder will change for git. Files can be
1. Added and commited (git is fully aware of them and all changes were registered)
2. Added and not commited (git is aware of the file, but there are changes not stored in git)
3. Untracked (git is not aware of the file)

**git status** allows us to see the state of the files in our repository. Now our only file should be untracked.

```python
git status # Check the status of your repository
```

Git starts tracking files when we first **add** them

```python
git add first_program.py #Start tracking the file
git status # Check how the status of your repository changed
```

Some times we want to add several files at the time, we can use

```python
git add -A # Add all files in the repository
git add . # Add all files in the current folder and its subfolders (not parent folders)
```

We can undo the add command if we staged a file by mistake. This doesn't remove the file, only tells git to stop tracking the last changes.

```python
git restore --staged filename # Unstage a file
```

Staged/added files are not really stored by git. Adding only serves to tell git what we actually want to save. **commit** is the command used to save snapshots of our project. The commit program **always** comes with a message specifying what were the changes from the last commit.

```python
git commit -m "Initial commit" # Commit the changes with a message
git status # Check how the status of the repository changed.
```

The history of commits is tracked, and we can always go back to a previous commit. To see the record of commits we use git log

![gitlog](Images/Screenshot_git_Log.png)

```python
git log # View the commit history
git log --stat # View what files were changed in each commit
git log --oneline # View the commit history in a compact format
```

Everytime we add changes to a file, the previous commit will get "outdated", meaning that new changes are not stored untill we add and commit them. So the workflow is: 

 Make changes in the project &rarr; add them with git add &rarr; commit them with git commit -m "commit message"

We can add and commit **already tracked files** with

```python
git commit -a -m "message" # add all modified files to the commit, but not new files
```

## Hands on I: learn pull, push, and merge.

Starting point:
Now you should have a local repository containing the Python program printing "Hello, World!" and an empty remote repository. The objective of this activity is learn to syncrhonize both repositories using **pull** and **push**.

1. *Add remote*: Connect your local and remote repositories and with **git remote add**. This command takes the URL of your remote repository and links it to your local. We need to provide a pseudonym for the remote repository, by convenction it is called "origin", but you could really use any name. This command will demmand your Github credentials to be completted.

```python
git remote add origin https://github.com/yourusername/your-repo.git 
```

2. *Fetch*: Once the remote is connected we have to tell git what is in there. We do so with **git fetch**, which takes the information in the remote *without* using it to update our local.

```python
git fetch origin # Fetch the latest changes from the remote repository
```

3. *diff*: We know that there are different files in the local and remote, let's check it with **git diff**. diff says what are the differences between two branches.  We didn't talk about branches so far, if you only used the commands explained so far no branches have been created and both the local and remote contain each only one branch. To be more complicated, both branches, the one in the local and the remote, are both named "main". We differentiate between the two because the branch in the remote is in "origin" (the remote's pseudonym), so it is called "origin/main" while the local one is simply "main".

```python
git diff --name-status origin/main main
```

4. *Push --Force*: Upload local files with **git push**. Using git push requires indicating again the branch in the local from where we extract the information and the branch of the remote where we want to upload the latest changes (both called "main"). Currently the local and remote repositories have different histories (both of them where created differently and contain different information). This history *divergence* makes that when we try to upload from the local to the remote, git would crash to avoid overwritting information. Since *we know* that the proper information is in the local while the remote is empty we can force overwritting with the option **--force**. **Warning:** git push --force should only be used in cases like this, to initialize the remote, in which we are EXTREMELLY sure that no important information can be overwritten by mistake.

```python
git push --force origin main:main # Push the changes to the remote repository "origin" from the local branch "main" to the remote branch "main"
```

5. *Using Github to edit*: Great! Now our git repository is synchronized with its remote version. You can check this running git diff again. Github is not only used to store, we can also use it to edit the code, create files or create folders. Also, other collaborators could ask for pushing their own changes to your repository. Overall, we could expect changes comming from both the local or the remote repositories and we have to know how to syncrhonize accordingly. Now go to the python program in the Github repository and edit it through the edit option (change the "Hello, World!" message for something else).
![Github_edit_file](Images/Screenshot_github_edit_file.png)

After editing, commit the changes (no need of adding when working from Github).
![Github_commit](Images/Screenshot_Github_commit.png)

5. *Pull* : Now is the local repository the one that is outdated, we can check this running first the fetch command (to update information without implementing changes) and then the diff command. We update the local repository through **pull**.

```python
git pull origin main:main
```

3. *shortcut*: since writting "git push origin main:main" everytime that we want to upload things to the remote is a bit tedious, we can tell git that all our "push" statements are of the type git push origin main:main using.

```python
git push -u origin main # Run only once.
```

After the above command is run, then running "git push" is the same as  running "git push origin main:main".
