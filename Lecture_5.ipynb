## Functions

Functions take arguments as inputs, include documentation (a docstring) explaining what they do, and return outputs.

```python
def function_name(arg1, arg2, arg3):
    '''
    Documentation: here we explain what the function does. 
    This can be 
    on multiple
    lines
    '''
    # Here we can write the code that performs the desired operations using arg1, arg2, and arg3.
    return a,b,c # Return the results of the operations. (Optional)
```

Example:

```python
def hello(name):
    """prints 'hello' followed by the arg name"""
    print("hello,", name)

hello("cat")
```

Arguments can be input by order or name (keyword argument).

```python
def hello(name1,name2):
    """prints 'hello' followed by two  names"""
    print("hello,", name1, "and", name2)

hello("cat", "dog")                 #Call the function with two positional arguments

hello(name2="cat", name1="dog")     #Call the function with two keyword arguments (order does not matter)
```

You can also define functions with default argument values. For example:

```python
def add_numbers(num1, num2=2):
    """Returns the sum of two numbers. Second number has a default value of 2."""
    return num1 + num2

print(add_numbers(1, 3))  # Output: 4
print(add_numbers(1))     # Output: 3 (uses default value for num2)
```

An argument can be mostly any Python object, including numbers, strings, lists, dictionaries, and even other functions.

```python
def mean(numbers):
    """Returns the mean of a list of numbers."""
    return sum(numbers) / len(numbers)

mean_value = mean([1, 2, 3, 4, 5])  # Output: 3.0
```

You can also return multiple values.

```python
def mean_and_errors(numbers):
    """Returns the mean and standard error of a list of numbers."""
    import numpy as np
    mean_value = np.mean(numbers)
    std_error = np.std(numbers) / np.sqrt(len(numbers))
    return mean_value, std_error

numbers = [1, 2, 3, 4, 5]
mean_value, std_error = mean_and_errors(numbers)
print(f"Mean: {mean_value}, Standard Error: {std_error}")
```

Functions can lack a return statement, in which case they return None by default.

```python
import matplotlib.pyplot as plt
import numpy as np
def Add_logscales_xy():
    """This function add logscales to the x and y axes of a plot.
    It has no arguments and does not return any value. It modifies the current plot in place."""
    plt.xscale('log')  # Set x-axis to logarithmic scale
    plt.yscale('log')  # Set y-axis to logarithmic scale

plt.figure(figsize=(10, 3))  # Create a figure with a specified size
xs = np.logspace(0.1, 2, 100)  # Generate 100 points between 10^0.1 and 10^2
ys = xs**2  # y = x^2
plt.subplot(1, 2, 1)  # Create a subplot
plt.plot(xs, ys)  # Plot the data
plt.title('Linear Scale')  # Set title for the subplot
plt.subplot(1, 2, 2)  # Create another subplot
plt.plot(xs, ys)  # Plot the data
Add_logscales_xy()  # Call the function to add log scales to the axes
plt.title('Log Scale')  # Set title for the subplot
```
