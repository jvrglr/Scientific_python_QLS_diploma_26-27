## Summary:
The geometric Brownian motion is a paradigmatic model used in many disciplines such as economics and ecology. It is a very interesting process to learn about stochastic processes since it is exactly solvable, yet it has counter-intuitive properties.

The model is defined in the Stratonovich sense (i.e., standard calculus)
$$\dot{x}(t) = \left[a +\sigma \xi(t)\right]x(t) $$
where $a$ and $\sigma$ are constants and $\xi(t)$ is a Gaussian white noise with $\langle \xi(t) \xi(t')\rangle =\delta (t-t')$. Furthermore, we consider the case in which $a<0$. The differential equation is exactly solvable:

$$x(t) = x(0)e^{a t+\sigma W(t)}.$$

Where $W(t)$ is the Wiener process. Since the solution to the SDE is known, all moments are also known

$$\langle x^n(t) \rangle= x^n(0) e^{a n t} e^{n^2 \sigma^2 t/2}.$$

**Apparent contradiction:** For $a<0$, trajectories tend to $x=0$ without ever touching it (as in the deterministic case $\sigma=0$). Using precise terms, as time goes to infinity we will find the process as close to zero as we want
$$\forall \epsilon\quad \lim_{t\to \infty} P[x(t)<\epsilon]=1.$$
Nevertheless, depending on the values of $a$ and $\sigma$ there is a threshold moment $n_{th}$ such that all higher moments diverge,
$$ \forall n>n_{th}\quad \lim_{t\to \infty} \langle x^n(t) \rangle=\infty.$$
Therefore, trajectories converge to zero but their moments diverge! How is that possible!?
