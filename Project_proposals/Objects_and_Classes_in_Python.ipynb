## Summary
In Python, everything is an object: numbers, lists, functions, NumPy arrays, pandas DataFrames, matplotlib figures. Classes are the tool to create your own types of objects, bundling data (attributes) together with the functions that act on them (methods). You have been using objects throughout the course without defining any: `arr.shape` is an attribute of a NumPy array, `df.mean()` is a method of a DataFrame, and `fig, ax = plt.subplots()` returns two objects.

This course focused on a function-based style of programming, and we did not have time to cover classes. In this project you will learn them on your own, teach them to your colleagues, and use them to build a reusable numerical tool.

## Numerical method: Runge Kutta
We want to solve systems of ODEs of the form

$$
\frac{d\mathbf{y}}{dt} = \mathbf{f}(t, \mathbf{y}), \qquad \mathbf{y}(t_0) = \mathbf{y}_0,
$$

where $\mathbf{y}$ can be a vector. With a time step $h$ and $t_n = t_0 + n h$, the fourth-order Runge–Kutta method (RK4) computes the solution at the next time as

$$
\begin{aligned}
\mathbf{k}_1 &= \mathbf{f}(t_n,\ \mathbf{y}_n), \\
\mathbf{k}_2 &= \mathbf{f}\!\left(t_n + \tfrac{h}{2},\ \mathbf{y}_n + \tfrac{h}{2}\,\mathbf{k}_1\right), \\
\mathbf{k}_3 &= \mathbf{f}\!\left(t_n + \tfrac{h}{2},\ \mathbf{y}_n + \tfrac{h}{2}\,\mathbf{k}_2\right), \\
\mathbf{k}_4 &= \mathbf{f}\!\left(t_n + h,\ \mathbf{y}_n + h\,\mathbf{k}_3\right), \\
\mathbf{y}_{n+1} &= \mathbf{y}_n + \frac{h}{6}\left(\mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4\right).
\end{aligned}
$$

The global error of RK4 scales as $h^4$. For comparison, the Euler method, $\mathbf{y}_{n+1} = \mathbf{y}_n + h\,\mathbf{f}(t_n, \mathbf{y}_n)$, has a global error that scales as $h$.

### Project main idea
* The solver must not know which model it is solving. Changing the model must not require changing the solver code.
* Models must be objects too: a base model class, and one subclass per model that stores its own parameters and defines $\mathbf{f}(t, \mathbf{y})$.
* The solver should store the results (times and solution), so that they can be used later, for example by a plotting method.
