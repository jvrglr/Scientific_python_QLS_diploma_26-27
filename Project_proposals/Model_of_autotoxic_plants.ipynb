## Summary
Some plants produce toxins that slow down their own growth (autotoxicity). In a recent paper we modeled this process with the following stochastic differential equations, interpreted in the **Itô** sense:

$$
\begin{cases}
\dfrac{dx}{dt} = x - x y + D \sqrt{x + x y}\, \xi(t), \\[2mm]
\dfrac{dy}{dt} = \rho (x - y),
\end{cases}
$$

where $x$ and $y$ are the densities of plants and toxins, and $\xi(t)$ is Gaussian white noise. The parameter $\rho$ is the ratio between the toxin decay rate and the plant growth rate (small $\rho$ means toxins degrade slowly), and $D$ is the strength of the demographic fluctuations, which decreases with system size.

In this model, extinction ($x = y = 0$) is the only long-term fate: sooner or later, every trajectory gets absorbed there. However, the path to extinction varies a lot between trajectories:
* **Short-lived**: the population dies before it can grow.
* **Excitable**: the population grows to a large maximum (boom), the accumulated toxin makes it crash (bust), and it dies during the crash.
* **Persistent**: the population survives the first crash and fluctuates around $x = y = 1$ for such long times that it is practically immortal.

The main result of the paper is that, starting without toxins ($y(0) = 0$), there are sharp transitions between these regimes: depending on $D$ and $\rho$, almost all trajectories follow one of the three pathways.

## Objective
Study numerically how the initial amount of toxin, $y(0)$, changes these probabilities. In particular: in the excitable regime, where plants starting without toxins die right after the boom, can some initial toxin increase their chances of survival?

## Numerical method: Milstein scheme
For an Itô SDE $dx = f(x,t)\,dt + g(x,t)\,dW(t)$, the Milstein scheme with time step $\Delta t$ reads

$$
x_{n+1} = x_n + f(x_n)\,\Delta t + g(x_n)\,\Delta W_n + \frac{1}{2}\, g(x_n)\, \frac{\partial g}{\partial x}(x_n) \left[(\Delta W_n)^2 - \Delta t\right],
$$

where $\Delta W_n = \sqrt{\Delta t}\, N(0,1)$ are independent Gaussian increments.

Two practical points:
* $x = 0$ is an absorbing state: if a step gives $x_{n+1} \le 0$, set $x = 0$ and stop updating $x$ for that trajectory. 
* stop updating trajectories once they are absorbed or classified!
