## Exercise 1 — Cleaning Temperature Measurements

The following temperatures were recorded during an experiment:

```python
temperature = [
    298.4, 299.1, 300.2, np.nan, 301.5,
    299.8, 300.6, 301.1, np.nan, 298.9,
    299.7, 300.4, 302.1, 301.6, np.nan,
    300.1, 299.5, 300.8, 301.3, 299.9
]
```

1. Use a `for` loop to iterate through the measurements.
2. Use `np.isnan()` and an `if` statement to identify missing measurements.
3. Store only the valid measurements in a new list called `clean_temperature`.
4. Count the number of missing measurements.
5. Calculate the mean temperature of the cleaned data using `def` using two decimal places.

Solution:

import numpy as np

# Temperature measurements
temperature = [
    298.4, 299.1, 300.2, np.nan, 301.5,
    299.8, 300.6, 301.1, np.nan, 298.9,
    299.7, 300.4, 302.1, 301.6, np.nan,
    300.1, 299.5, 300.8, 301.3, 299.9
]

# Empty list to store valid temperature measurements
temp = []

# Counter for missing measurements
count = 0

# Loop through the temperature data with index and value
for i, val in enumerate(temperature):

    # Check if the measurement is NaN
    if np.isnan(val):
        print(i, val)
        count += 1

    # Store valid measurements
    else:
        temp.append(val)

# Display cleaned data, number of valid measurements, and data type
print(temp, len(temp), type(temp))

# Display number of missing measurements
print(count)


# Function to calculate the mean temperature
def mean_temp(temp):
    mean_temp = sum(temp) / len(temp)
    return mean_temp


# Calculate and display the mean of the cleaned data
print(mean_temp(temp))
