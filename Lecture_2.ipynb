### Variables

Variables are **symbols that refer to objects**.

#### Variable Names

> - Variable names are **case-sensitive**.
> - May contain: `a-z`, `A-Z`, `0-9`, `_`.
> - Cannot start with a number.
> - Python does not impose a length limit on variable names.

#### Naming Conventions

> - Variables and functions are written in **lowercase**.
> - Use **snake_case** for multiple words: `num_particles`.
> - Classes normally start with a **capital letter**: `Particle`
> - A leading underscore `_name` indicates an **internal/non-public** name.
> - Double underscores `__name__` are reserved for **special Python-defined names**.
> - Avoid using names of built-in functions: `print`, `list`, `sum`, `max`, etc.
> - Python **keywords** cannot be used as variable names: `for`, `if`, `class`, `return`, etc.

#### Assignment

A single assignment:

`x = 10`

Multiple assignment:

`a, b, c = 1, 1.0, "ICTP"`

#### Statements

A semicolon `;` can separate multiple statements on the same line:

`a, b, c = 1, 1.0, "ICTP"; print(c)`

However, one statement per line is preferred for readability.

#### Dynamic and Implicit Typing

In Python, variables are created **dynamically and implicitly**.

> - **Implicit:** the type does not need to be explicitly declared as in C or Fortran.
> - **Dynamic:** a variable name can refer to objects of different types during execution.
> - **Duck typing:** Python generally determines whether an object is suitable from the operations it supports rather than requiring a specific declared type.

### Operator Precedence

| Order | Operator | Operation | Direction |
|---|---|---|---|
| **1 — Highest** | `()` | Grouping / Parentheses | Innermost first |
| **2** | `**` | Exponentiation | **Right → Left** |
| **3** | `+x`, `-x` | Unary plus, unary minus | Right → Left |
| **4** | `*`, `/`, `//`, `%` | Multiplication, division, floor division, modulus | **Left → Right** |
| **5 — Lowest** | `+`, `-` | Addition, subtraction | **Left → Right** |

#### NOTES:

> Grouping `()` is evaluated first.

> Operators with the same precedence are generally evaluated **left to right**.

> Multiple exponentiation `**` e.g. 2**3**4 is an exception. It is evaluated from **right to left**.

> Nested parentheses are evaluated from the **innermost to the outermost**.

````python
x = 2 + 2 # This returns an integer
print(x)

x = 50 - 5*6 # This returns an integer
print(x)

x = (50 - 5*6) / 4 # Division always returns a floating point (float) number
print(x)
````

````python
x = 8 / 5  # basic division
print(x)

x = 17 // 3  # floor division discards the fractional part
print(x)

x = 17 % 3  # the % operator returns the remainder of the division
print(x)
````



## Strings

Strings (`str`) are ordered sequences of characters.

> - Can be defined using single quotes `'...'` or double quotes `"..."` with the same result.  
>   Example: `'ICTP'` or `"ICTP"`
>
> - **Immutable** — cannot be changed after creation.
>
> - **Iterable** — characters can be accessed one at a time, for example using a `for` loop.  
>   Example: `"ICTP"` → `I`, `C`, `T`, `P`
>
> - **Ordered** — characters maintain their position and sequence.  
>   Example: `"ICTP"` → `I`, `C`, `T`, `P`
>
> - **Indexed** — characters can be accessed directly by their position (index).  
>   Indexing starts from `0`.  
>   Example: `"ICTP"[0]` → `"I"`

### String Indexing

An **index (`idx`)** represents the position of a character in a string.

> - Positive indexing starts from `0` at the beginning.
> - Negative indexing starts from `-1` at the end.

For the string `"ICTP"`:

| Character | `I` | `C` | `T` | `P` |
|---|---:|---:|---:|---:|
| Positive index | `0` | `1` | `2` | `3` |
| Negative index | `-4` | `-3` | `-2` | `-1` |

Examples:

`"ICTP"[0]` → `"I"`  
`"ICTP"[1]` → `"C"`  
`"ICTP"[-1]` → `"P"`  
`"ICTP"[-2]` → `"T"`

### String Slicing

`str[start_idx:end_idx:stride]`

| Parameter | Meaning | Example |
|---:|---:|---:|
| `start_idx` | Start at this index and continue to the end | `"ICTP"[1:]` → `"CTP"` |
| `end_idx` | Start at the beginning and stop before this index | `"ICTP"[:2]` → `"IC"` |
| `stride` | Select characters at a specified step | `"ICTP"[::2]` → `"IT"` |


#### NOTES:

> `start_idx` is included, while `end_idx` is excluded.

````python
print('ciao')  # single quotes

print('doesn\'t')  # use \' to escape the single quote...

print("doesn't")  # ...or use double quotes instead


print('"Yes," they said.')

print("\"Yes,\" they said.")

print('"Isn\'t," they said.')
````

## Copying variables

In Python (and many other languages), it is important to understand the difference between assigning a copy using ```=```, a **shallow copy**, and a **deep copy**. 

- The operator ```=``` points your new variable to the existing object, it does not create a new oject. 
- A shallow copy creates a new object that contains references to the original object. 
- A deep copy creates a new object that contains copies of the original object, there are no references to the original object. 

To demonstrate the difference, we can use the operator ```id()``` to check the identity of each type of copy. The output of this operator will tell us whether the object within a copy contains a reference to the original  

First, we will use the operator ```=```.

````python
a = [[0,1,2], [0,2,4]]
b = a

print(id(a) == id(b))          # True = b is the same object as a
print(id(a[0]) == id(b[0]))    # True = b[0] is the same object as a[0]

````

Next, we will perform a shallow copy.

````python
import copy # we will need this module to perform shallow and deep copies

b = copy.copy(a)

print(id(a) == id(b))          # False = b is now a new object
print(id(a[0]) == id(b[0]))    # True = b[0] is the same object as a[0]
````

Finally, a deep copy.

````python
b = copy.deepcopy(a)

print(id(a) == id(b))          # False = b is now a new object
print(id(a[0]) == id(b[0]))    # False - b[0] is now a new object
````
